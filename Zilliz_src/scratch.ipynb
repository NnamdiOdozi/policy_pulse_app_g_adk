```python
from typing import Any
```

```python
def list_words(words: list[str]) -> None:
    for i, word in enumerate(words):
        print(i, word)
```

```python
class ListWords:
    def __call__(self, words: list[str]) -> Any:
        for i, word in enumerate(words):
            print(i, word)
```

```python
def main() -> None:
    words = ["code", "python", "ai", "refactor", "bug"]
    list_words_fn = ListWords()
    list_words_fn(words)

if __name__ == "__main__":
    main()
```
